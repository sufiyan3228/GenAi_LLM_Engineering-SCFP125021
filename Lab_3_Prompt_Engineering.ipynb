### Exercise 1 — Doctor Summary Generator

Build a prompt that takes a patient's raw consultation notes and generates a clean doctor's summary.

**Input:** `patient_notes`, `patient_name`

**Output:** a summary with fixed sections — Symptoms, Diagnosis, Recommendation — in the same format every time, regardless of input.

```python
import os
from dotenv import load_dotenv
from groq import Groq
```

```python
load_dotenv()
```

```python
user_prompt_1 = "The patient name is {patient_name}. the patient notes are {patient_notes}"
patient_name = input("Enter patient name : ")
patient_notes = input("Enter patient notes : ")
user_message_1 = user_prompt_1.format(patient_name=patient_name, patient_notes=patient_notes)
```

```python
message_1 = [

    {"role": "system",
         "content": "You are an Assitant to the Medical Doctor, who takes patient's raw consultation notes and then generates a clean doctor's summary. Let the Summary be concise yet all inclusive of the datails provided. a summary with fixed sections — Symptoms, Diagnosis, Recommendation — in the same format every time, regardless of input.. Be strict and to the point with the summary."
    },

    {"role": "user",
     "content": user_message_1
    }
]
```

```python
client = Groq(api_key=os.environ["GROQ_API_KEY"])
```

```python
chat_completion = client.chat.completions.create(messages=message_1, model="openai/gpt-oss-20b")
response = chat_completion.choices[0].message.content
print(response)
```

### Exercise 3 — Resume Field Extractor

Build a prompt that takes resume text and pulls out only the requested fields.

**Input:** `resume_text`, `fields_to_extract`

**Output:** strict JSON containing only the requested fields, nothing else

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser
```

```python
llm=ChatGroq(
    model= "openai/gpt-oss-20b" ,
    api_key=os.environ["GROQ_API_KEY"]
)
```

```python
json_parser=JsonOutputParser()
```

```python
resume_text = input("Enter the resume text : ")
field_names = input("Enter the fields to be extracted : ")
```

```python
print(resume_text),
print(field_names)
```

```python
resume_input=ChatPromptTemplate.from_messages([
    ("system", "You are an Assitant to an Hiring Manager, who takes the complete {resume_text} as an input and then asks user for the required fields, then only extract and output the fields to be extracted in a JSON format. ."),
    ("user", "Fields to be extracted {field_names}")
]).partial(JSON_FORMATING=json_parser.get_format_instructions())
```

```python
resume_input
```

```python
input=resume_input.invoke({"resume_text": resume_text, "field_names": field_names})
```

```python
response=llm.invoke(input)
```

```python
response.content
```

```python
parsed_response=json_parser.parse(response.content)
```

```python
parsed_response
```

### Exercise 4 — Customer Support Reply Generator

Build a prompt that generates a support reply to a customer message.

**Input:** `customer_message`, `company_name`, `max_words`

**Output:** a reply that stays within the word limit, with consistent tone across different inputs

```python
llm_input=ChatGroq(
    model= "openai/gpt-oss-20b" , 
    api_key=os.environ["GROQ_API_KEY"]
)
```

```python
customer_message = input("Please Enter your message: ")
company_name = input("Company name: ")
max_words = 30
```

```python
support_invoke=ChatPromptTemplate.from_messages([
    ("system", "You are an Customer support reply generator, who takes the complete {customer_message} as an input and then takes {company_name} from the user, and generates the reply in {max_words} only."),
    ("user", "My Query is {customer_message} regarding the company {company_name}. Please generate a reply in {max_words} words only.")
])
```

```python
input_query=support_invoke.invoke({"customer_message": customer_message, "company_name": company_name, "max_words": max_words})
```

```python
llm_response=llm_input.invoke(input_query)
```

```python
llm_response.content
```

```python
word_count = len(llm_response.content.split())
print("max words :", max_words)
print("word count :", word_count)
```

### Exercise 5 — Multi-language Translator

Build a prompt that translates a sentence into a target language at a given formality level.

**Input:** `sentence`, `target_language`, `formality`

**Output:** a translated sentence matching the requested formality level

```python
llm_input=ChatGroq(
    model= "openai/gpt-oss-20b" , 
    api_key=os.environ["GROQ_API_KEY"]
)
```

```python
sentence = "AI will define the new era of technology & Innovation"
target_language = ("Hindi", "Marathi")
formality = "Casual & Straight to the point"
```

```python
translate_input=ChatPromptTemplate.from_messages([
    ("system", "You are an Language Translator, who takes the complete {sentence} as an input and then takes {target_language} from the user, and generates the translation in {formality} only."),
    ("user", "I want the {sentence} to be translated to {target_language}. Please generate a translation in {formality} only.") 
])    
```

```python
translation_temp=translate_input.invoke({"sentence": sentence, "target_language": target_language, "formality": formality})
```

```python
translation_resp=llm_input.invoke(translation_temp)
```

### Exercise 6 — Iterative Content Refiner

Build a prompt that takes a draft and revises it to fix one specific issue, across multiple rounds.

**Input:** `draft_text`, `issue_to_fix`

**Output:** a revised version of the text that resolves that specific issue, run for at least 3 rounds

```python
llm_input=ChatGroq(
    model= "openai/gpt-oss-20b" , 
    api_key=os.environ["GROQ_API_KEY"]
)
```

```python
draft_text = ("the growing rapidly pollution because factories many waste harmful releasing are environment the into, government strict should rules introduce but companies the profit their about only thinking are.")
issue_to_fix= ("Grammatical errors")
```

```python
refiner_input=ChatPromptTemplate.from_messages([
    ("system", "You are an Text Refiner, who takes the complete {draft_text} from user and improves it according to {issue_to_fix}."),
    ("user", "Fix the follwing {draft_text} with respect to {issue_to_fix}.")
])
```

```python
refine_temp=refiner_input.invoke({"draft_text": draft_text, "issue_to_fix": issue_to_fix})
```

```python
refined_resp = draft_text

for i in range(1, 4):
    print(f"Attempt {i}:")

    refined_resp = llm_input.invoke(refine_temp)

    print(refined_resp.content)
```
